```python
# libraries
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
 
# create data
size = 100000
df = pd.DataFrame({
   'x': np.random.normal(size=size),
   'y': np.random.normal(size=size)
})
```

```python
fig, axs = plt.subplots(nrows=2, ncols=2, figsize=(8,8))

# Big bins
axs[0,0].hist2d(df['x'], df['y'], bins=(50, 50), cmap=plt.cm.jet)
axs[0, 0].set_title('bins = (50, 50)')
 
# Small bins
axs[0,1].hist2d(df['x'], df['y'], bins=(600, 600), cmap=plt.cm.jet)
axs[0, 1].set_title('bins = (600, 600)')
 
# If you do not set the same values for X and Y, the bins won't be a square!
axs[1,0].hist2d(df['x'], df['y'], bins=(600, 30), cmap=plt.cm.jet)
axs[1, 0].set_title('bins = (600, 30)')

# If you do not set the same values for X and Y, the bins won't be a square!
axs[1,1].hist2d(df['x'], df['y'], bins=(30, 600), cmap=plt.cm.jet)
axs[1, 1].set_title('bins = (30, 600)')

plt.show()
```

Map with markers with Python and Folium

Modified the code found here: https://python-graph-gallery.com/312-add-markers-on-folium-map/

to show makers of 10 Canadian provinces.

```python
# import the library
import folium

# Data for 10 Canadian provinces (capital cities)
data = pd.DataFrame({
    'lon': [
        -63.5752,   # Halifax
        -66.6431,   # Fredericton
        -63.1311,   # Charlottetown
        -52.7126,   # St. John's
        -79.3832,   # Toronto
        -73.5673,   # Montreal
        -97.1384,   # Winnipeg
        -106.6700,  # Regina
        -113.4909,  # Edmonton
        -123.1207   # Victoria
    ],
    'lat': [
        44.6488,
        45.9636,
        46.2382,
        47.5615,
        43.6532,
        45.5017,
        49.8951,
        50.4452,
        53.5461,
        48.4284
    ],
    'name': [
        'Nova Scotia',
        'New Brunswick',
        'Prince Edward Island',
        'Newfoundland and Labrador',
        'Ontario',
        'Quebec',
        'Manitoba',
        'Saskatchewan',
        'Alberta',
        'British Columbia'
    ]
})

# Create map centered on Canada
m = folium.Map(location=[56.1304, -106.3468], tiles="OpenStreetMap", zoom_start=4)

# Add markers
for i in range(len(data)):
    folium.Marker(
        location=[data.iloc[i]['lat'], data.iloc[i]['lon']],
        popup=data.iloc[i]['name'],
        icon=folium.Icon(color="red")
    ).add_to(m)

# Show map
m
```
