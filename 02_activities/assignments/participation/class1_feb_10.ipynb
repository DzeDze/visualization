```python
# Import libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
import warnings
warnings.filterwarnings("ignore")
```

```python
np.random.seed(613)

# random values for x and y
x = np.arange(50)
y = np.random.randint(0, 100, 50)

# fig of (5,3) inches size, ax for plotting
fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x, y)
```

```python
fig, ax = plt.subplots(figsize=(5,3))
ax.bar(x,y)
```

```python
fig, ax = plt.subplots(figsize=(5,3))
ax.barh(x,y)
```

```python
fig, ax = plt.subplots(figsize=(5,3))
# line chart
ax.plot(x,y)
```

```python
fig, ax = plt.subplots(figsize=(5,3))
# histogram
ax.hist(y)
```

```python
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y)
ax.set_title('Total growth over time')
ax.set_ylabel('Total growth')
ax.set_xlabel('Years since start')
fig.tight_layout()
```

```python
font1 = {'family':'sans-serif', 'color':'blue', 'size':20}
font2 = {'family':'monospace', 'color':'green', 'size':14}

fig, ax = plt.subplots(figsize=(5,3)) 
ax.plot(x,y)

ax.set_title('Total growth over time', fontdict=font1, loc = 'left')
ax.set_ylabel('Total growth', fontdict=font2)
ax.set_xlabel('Years since start', fontdict=font2)
fig.tight_layout()
```

```python
fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x, y, marker='*', color = "indigo")
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y,marker= '*', color = 'indigo', linestyle = '--', linewidth = 2)
```

```python
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y,
        marker='*', #marker appearance, can be: *, o, x, etc.
        markersize = 12, 
        color = '#7425b9', 
        linestyle = '--', 
        linewidth = 2, 
        markeredgecolor = '#fa9359', 
        markerfacecolor = '#000000' )
ax.grid() 
```

```python
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(
    x,y,
    marker='*', 
    markersize = 12, 
    color = '#7425b9', 
    linestyle = '--', 
    linewidth = 2, 
    markeredgecolor = '#fa9359',
    markerfacecolor = '#000000' 
)

ax.grid(
    color = "#25b94c", 
    linestyle = 'dotted', 
    linewidth = 2, 
)
```
